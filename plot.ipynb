```python
import matplotlib.pyplot as plt
import numpy as np

freq_ML = np.genfromtxt('freq_ML')
freq_xTB = np.genfromtxt('freq_xTB')

plt.plot(freq_xTB[0:3],freq_ML[0:3],'.',label = r'H$_\mathrm{2}$')
plt.plot(freq_xTB[3:8],freq_ML[3:8],'.',label = r'N$_\mathrm{2}$')
plt.plot(freq_xTB[8:13],freq_ML[8:13],'.',label = r'O$_\mathrm{2}$')
plt.plot(freq_xTB[13:18],freq_ML[13:18],'.',label = r'F$_\mathrm{2}$')
plt.plot(freq_xTB[18:23],freq_ML[18:23],'.',label = r'CO')
plt.plot(freq_xTB[23:28],freq_ML[23:28],'.',label = r'HF')

plt.legend()
plt.plot([0,10000],[0,10000],'k')

plt.xlim([0,5000])
plt.ylim([0,5000])

plt.ylabel(r'$\tilde{\nu}_\mathrm{ML}~[\mathrm{cm^{-1}}]$')
plt.xlabel(r'$\tilde{\nu}_\mathrm{xTB}~[\mathrm{cm^{-1}}]$')

plt.savefig('Diatomic_ML_model.svg')
plt.savefig('Diatomic_ML_model.png')
plt.savefig('Diatomic_ML_model.eps')

plt.show()
```
