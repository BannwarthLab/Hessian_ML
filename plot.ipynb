```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 unused import
import matplotlib.pyplot as plt
import numpy as np
from scipy import linalg
from operator import matmul

plt.style.use('rwth_colors.mplstyle')

def rot_Z(alpha):
    R = np.array([[np.cos(alpha), -np.sin(alpha), 0],
                   [np.sin(alpha),  np.cos(alpha), 0],
                   [0,          0,                 1]])
    return R

def rot_X(alpha):
    
    R = np.array([[1,             0,              0],
                   [0, np.cos(alpha), -np.sin(alpha)],
                   [0, np.sin(alpha), np.cos(alpha)]])
    return R 


def angle_two_vec(a,b):
    if linalg.norm(a) == 0. or linalg.norm(b) == 0.:
            cosangle = 0
    else:
        cosangle = matmul(a,b)/linalg.norm(a)/linalg.norm(b)
    
    angle = np.arccos(np.clip(cosangle,-1,1))
    return angle

ax = plt.figure().add_subplot(projection='3d')

x = np.array([0,0,0])
y = np.array([0,0,0])
z = np.array([0,0,0])



u = np.array([1,0,0])
v = np.array([0,1,0])
w = np.array([0,0,1])


a = np.linspace(0,np.pi*2,2001)

c_y = np.cos(a)
c_x = np.sin(a)
c_z = a*0

c = np.array([c_x,c_y,c_z]).T

axes = np.array([u,v,w])

dipm = np.array([0.3,0.2,-0.2]) * linalg.norm(np.array([0.3,0.2,-0.2]))


vec_z = np.array([-0.33,-0.31,0.37]) * linalg.norm(np.array([-0.33,-0.31,0.37]))

vec_x = np.cross(dipm,vec_z)


L1 = np.cross(vec_z,axes[2])
L1 = linalg.norm(L1)**-1*L1

L2 = np.cross(axes[2],vec_z)
L2 = linalg.norm(L2)**-1*L2

LL = np.cross(vec_z,axes[2])

alpha = angle_two_vec(LL,axes[0])
beta = angle_two_vec(vec_z,axes[2])
gamma = angle_two_vec(LL,vec_x)

#Find right rotation angle
if linalg.det(np.array([axes[0],axes[2],LL])) < 0.:
    alpha = 2*np.pi - alpha

if linalg.det(np.array([LL,vec_x,vec_z])) > 0.:
    gamma = 2*np.pi - gamma

R1 = rot_Z(gamma)
R2 = rot_X(beta)
R3 = rot_Z(alpha)

R = np.matmul(np.matmul(R1,R2),R3)

axes_rot = np.matmul(np.transpose(R),axes)

for i in range(len(c)):
    c[i] = np.matmul(np.transpose(R),c[i])
    
c=c.T

ax.quiver(x,y,z,axes[0],axes[1],axes[2],color = 'k',arrow_length_ratio=0.15)



ax.quiver(x,y,z,axes_rot[0],axes_rot[1],axes_rot[2],color = 'b',arrow_length_ratio=0.15)

print(axes_rot[0],axes[2])

ax.quiver(x,y,z,L1[0],L1[1],L1[2],color = 'red',arrow_length_ratio=0.0)
ax.quiver(x,y,z,L2[0],L2[1],L2[2],color = 'red',arrow_length_ratio=0.0)

L1 = L1*1.3
L2 = L2*1.2
ax.text(L1[0],L1[1],L1[2],r'$\mathrm{L_b}$',color = 'r')
ax.text(L2[0],L2[1],L2[2],r'$\mathrm{L_a}$',color = 'r')

ax.scatter(c_x,c_y,c_z,color = 'k',s = 0.1)
ax.scatter(c[0],c[1],c[2],color = 'b',s = 0.1)

ax_lim = 1.2

ax.set_xlim3d(-ax_lim, ax_lim)
ax.set_ylim3d(-ax_lim, ax_lim)
ax.set_zlim3d(-ax_lim, ax_lim)


ax.text(1.1, 0, 0, "x", color='k')
ax.text(0, 1.1, 0, "y", color='k')
ax.text(0, 0, 1.1, "z", color='k')
dist = 1.1 
axes_rot = axes_rot * dist

ax.text(axes_rot[0][0]*1.1,axes_rot[1][0]*1.1,axes_rot[2][0],r"x'",color = 'blue')
ax.text(axes_rot[0][1],axes_rot[1][1],axes_rot[2][1],r"y'",color = 'blue')
ax.text(axes_rot[0][2],axes_rot[1][2],axes_rot[2][2],r"z'",color = 'blue')




ax.set_axis_off()

plt.savefig('EulerAngle.svg')
plt.savefig('EulerAngle.eps')
```
